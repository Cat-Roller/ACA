Problem 1: The Matrix Architect
Task: 1. Create a 2D NumPy array with 5 rows and 3 columns filled entirely with the number 7. 2. Change the element in the third row, second column to 10. 3. Print the following attributes of the array: ndim, shape, and dtype.

```python
import numpy as np
matrix = np.full((5,3), 7)
matrix[2,1] = 10
print(matrix.ndim)
print(matrix.dtype)
print(matrix.shape)
```
