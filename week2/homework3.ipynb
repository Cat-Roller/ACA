<h2>First Task</h2>
<h5>
  Task: Loop through the numbers 1 to 20.

  If the number is divisible by 5, pass (do nothing).

  If the number is 17, break the loop entirely.

  Otherwise, print the number.
  </h5>

```python
for n in range(20):
    if n%5==0:continue
    if n == 17:break
    print(n)
```

<h2>Second Task</h2>
<h5>
    Task: You have a list: items = ["apple", "remove", "banana", "remove", "cherry"]. Use a while loop to find every instance of the string "remove" and delete it from the list.

    Hint: Be careful! Deleting items while looping with indices can be tricky.
  </h5>

```python
items = ['apple', 'remove', 'banana', 'remove', 'cherry']

while 'remove' in items:
    items.remove('remove')
print(items)
```

<h2>Third Task</h2>
<h5>
    Task: Given a list of mixed numbers data = [10, -3, 25, -9, 0, 12], create a new list that contains only the positive numbers, but in reverse order from how they appeared in the original list.
  </h5>

```python
data = [10, -3, 25, -9, 0, 12]
positive_reversed = []

for num in data[::-1]:
    if num > 0:
        positive_reversed.append(num)

print(positive_reversed)
```

<h2>Forth Task</h2>
<h5>
    Task: Create a new list that contains only the elements from my_list = ["a", "b", "c", "d", "e", "f"] that are located at even index positions (0, 2, 4...). You must use a for loop and range().  
  </h5>

```python
my_list = ['a', 'b', 'c', 'd', 'e', 'f']
even_letters = []

for i in range(0,len(my_list),2):
    even_letters.append(my_list[i])

print(even_letters)
```

<h2>Fifth Task</h2>
<h5>
    Task: A user provides prices one by one. If they enter a negative price, it's an error—use continue to skip it. If they enter 0, stop the input. Store the valid prices in a list and, at the end, print the list and the total sum.
  </h5>

```python
prices = []

while True:
    str_price = input('Enter the price===>')
    try:
        price = int(str_price)
        if price > 0:
            prices.append(price)
        elif price == 0:
            break
        else:
            print('Error:No negative prices')
    except ValueError:
        print('Sorry, the price has to be a positive number(to exit enter \'0\')')
print('Your list is:','='*50,prices,'='*50,sep='\n')
print(f'And the sum of all prices is {sum(prices)}')
```

<h2>Bonus Task</h2>
<h5>
  The Scenario: You are given a list of heights: altitudes = [100, 120, 110, 130, 150, 140, 160]. The Task: Find all "peaks." A peak is a number that is greater than the number before it AND greater than the number after it.

  Store these peaks in a new list.

  Logic Constraint: You cannot use built-in functions like max(). You must use a loop and index comparisons.
  </h5>

```python
altitudes = [100, 120, 110, 130, 150, 140, 160]
peaks = []

for i in range(1,len(altitudes)-1):
    if altitudes[i-1]<altitudes[i]>altitudes[i+1]:
        peaks.append(altitudes[i])

print(peaks)
```
