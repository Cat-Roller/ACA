Get a number betwen 1 - 100

if the number is less than 10 <= then print the x^e
if the number is betweenb 10 and 90 then print the log 2  of the x 
if the number is between >= 90 and 100 then print the square root of the number 


Return all the nubers rounded by 3 digits
all the other cases shall print(invalid number)

```python
import math
import random
```

```python
num = random.uniform(1,100)
if num<=10:
    print(f'The x^e of {num} is ',round(num**math.e,3))
elif 10<num<90:
    print(f'The log 2 of {num} is ',round(math.log2(num),3))
else:
    print(f'The square root of {num} is ',round(math.sqrt(num),3))
```
