<h2>Second Task</h2>
<h4>
Problem 2: The Number Detective Ask the user for a number.

If the number is greater than 0, print "This is a positive number."

If the number is less than 0, print "This is a negative number."

If the number is exactly 0, print "This is zero."
  </h4>

```python
num = int(input('Enter your number please'))
if num > 0:
    print('Your number is positive')
elif num == 0:
    print('Your number is exactly 0')
else:
    print('Your number is negative')
```

    Use the modulo operator % to find out if the number is even or odd.

<h2>Forth Task</h2>
<h4>
    Problem 4: The Math Bot Build a program that asks for two whole numbers.

    Then, ask the user: "Type 1 to Add or 2 to Subtract."

If they type 1, print the sum of the two numbers.

If they type 2, print the difference.

If they type anything else, print "Error: Invalid Choice."
  </h4>

```python
first_value = int(input('Enter your first number'))
second_value = int(input('Enter your second number'))
operator = input('Enter 1 to add previous numbers and 2 to substract')
if operator == '1':
    print(f'The sum of your numbers is {first_value+second_value}')
elif operator == '2':
    print(f'The difference between your numbers is {first_value-second_value}')
else:
    print('Error:invalid operator (for addition enter 1 for substraction 2)')
```
