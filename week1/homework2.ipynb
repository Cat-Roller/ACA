<h2>Second Task</h2>
<h4>
    Question 2: The Savings Account (Augmented Assignments)

You start with a bank balance of $1000. Write a program that simulates a deposit and a withdrawal using augmented assignments (+=, -=).

Create a variable balance = 1000.

Ask the user: "How much do you want to deposit?" Add this to the balance using +=.

Ask the user: "How much do you want to withdraw?" Subtract this from the balance using -=.

Print the final remaining balance.
  </h4>

```python
balance = 1000
dep=-1
draw=-1
while dep<0 or draw<0:
    dep = int(input('How much would you like to deposit'))
    draw = int(input('How much would you like to withdraw'))
    if dep<0 or draw<0:
        print('Sorry you cant input negative numbers')
balance+=dep
balance-=draw
print(f'Your remaining balance is {balance}')
```

    Question 3: String Slicer
Write a program that asks the user to input a single word (e.g., "Python").

Print the first 2 characters of the word.

Print the last 2 characters of the word.

Print the word in ALL UPPERCASE.

Print the length of the word.

  </h4>

```python
word= input('Entr your word')
while len(word)<2 or " " in word:
    print('Sorry the word has to have at least 2 letters and no whitespaces')
    #input validation can be improved
    word = input('Enter your word')

print(f'The fist two letters of your word are {word[:2]}')
print(f'The last two letters of your word are {word[-2:]}')
print(f'You word in uppercase is {word.upper()}')
print(f'The length of the word is {len(word)}')
```
