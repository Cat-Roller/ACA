Problem 1: The Volatile Stock Market
You have an array of 60 days of stock prices. However, on some days, the stock exchange was closed or the data was not recorded, resulting in NaN values.

Create an array of 60 values using np.random.uniform(150, 200, 60).

Randomly replace 10 of these values with np.nan.

Calculate the average stock price ignoring the NaN values.

Replace all NaN values with the average price you calculated in step 3.

Use a ufunc to calculate the daily price change (log return) using np.log().

```python
import numpy as np
prices = np.random.uniform(150, 200, 60)
prices[np.random.randint(0, 60, 10)] = np.nan
prices[np.isnan(prices)] = np.nanmean(prices)
print(np.log(prices))
```

Problem 5: Global Shipping Logistics
English: You have 80 packages. You need to calculate shipping costs based on weight and distance.

Create two 1D arrays of 80 elements: weights (1kg to 50kg) and distances (10km to 1000km).

The base shipping formula is: 
cost = (weight * 0.5) + (distance *0.1)
. Calculate this using ufuncs. Due to fuel surcharges, any distance over 500km gets an extra $20 flat fee.

Use np.where to apply this logic.
Calculate the standard deviation of the shipping costs.
 
Identify all packages that cost more than 1.5 standard deviations above the mean cost (High-Cost Deliveries).

```python
weights = np.random.uniform(1, 50, 80)
distances = np.random.uniform(10, 1000, 80)

cost = (weights * 0.5) + (distances *0.1)
cost = np.where(distances > 500, cost + 20, cost)
print(cost[cost > (cost.mean() + 1.5 * cost.std())])
```
